# パーティションのサイズ

このJupyter Notebookはパーティションのサイズを考える際に利用したものです。

```python
DISK_SIZE = 48

BOOT = 2
SWAP = 4
HOME = 10
VAR = 4
SRV = 4
TMP = 4
VAR_LOG = 2


def calc_remaining_root():
    return DISK_SIZE - BOOT - SWAP - HOME - VAR - SRV - TMP - VAR_LOG


print("Remaining for root:", calc_remaining_root())
```
